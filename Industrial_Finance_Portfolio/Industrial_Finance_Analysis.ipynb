# Industrial Manufacturing Finance & Profitability Analysis

## tl;dr
Analyze a synthetic industrial machinery manufacturer at sale-order level. Focus: revenue, gross profit, product profitability, order exposure, and material-cost sensitivity.

## Context & Methods
Source: `Industrial_Manufacturing_Finance_Project.xlsx`. Grain: one row per sale order. Figures are synthetic portfolio data.

```python
import pandas as pd
import matplotlib.pyplot as plt
sales=pd.read_excel('Industrial_Manufacturing_Finance_Project.xlsx', sheet_name='Sales_Orders')
sales['SO_Date']=pd.to_datetime(sales['SO_Date'])
sales['Month']=sales['SO_Date'].dt.to_period('M').astype(str)
sales.head()
```

```python
revenue=sales['Revenue'].sum()
gross_profit=sales['Gross_Profit'].sum()
print(f'Revenue: ₹{revenue:,.0f}')
print(f'Gross Profit: ₹{gross_profit:,.0f}')
print(f'Gross Margin: {gross_profit/revenue:.1%}')
```

## Results
### Product profitability

```python
product=sales.groupby('Model').agg(Revenue=('Revenue','sum'),Units=('Qty','sum'),Gross_Profit=('Gross_Profit','sum'))
product['Gross_Margin']=product['Gross_Profit']/product['Revenue']
product.sort_values('Gross_Profit',ascending=False)
```

```python
monthly=sales.groupby('Month').agg(Revenue=('Revenue','sum'),Gross_Profit=('Gross_Profit','sum'))
monthly['Gross_Margin']=monthly['Gross_Profit']/monthly['Revenue']
monthly['Revenue'].plot(kind='bar',figsize=(10,5),title='Monthly Revenue')
plt.ylabel('Revenue (₹)'); plt.tight_layout(); plt.show()
```

### Cost sensitivity
Test the effect of a 10% increase in material cost.

```python
scenario_gp=revenue-sales['Material_Cost'].sum()*1.10-sales['Labour_Cost'].sum()-sales['Overhead'].sum()
print(f'Gross Margin after +10% material cost: {scenario_gp/revenue:.1%}')
```
